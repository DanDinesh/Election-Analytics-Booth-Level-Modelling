```python
import re

def clean_text(value):

    if pd.isna(value):
        return ""

    value = str(value).upper().strip()

    # Replace comma with space
    value = value.replace(",", " ")

    # Remove unwanted symbols
    value = re.sub(
        r"[^A-Z0-9&() ]+",
        " ",
        value
    )

    # Remove multiple spaces
    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()

```

```python
def canonical_party(value):

    original = clean_text(value)

    if original in ["", "0", "NAN"]:
        return "UNKNOWN"

    compact = re.sub(
        r"[^A-Z0-9]+",
        "",
        original
    )

    # INC
    if "INDIANNATIONALCONGRESS" in compact:
        return "INC"

    # AIADMK - check before DMK
    if "ALLINDIAANNADRAVIDAMUNNETRAKAZHAGAM" in compact:
        return "AIADMK"

    if "ANNADRAVIDAMUNNETRAKAZHAGAM" in compact:
        return "AIADMK"

    # DMK
    if "DRAVIDAMUNNETRAKAZHAGAM" in compact:
        return "DMK"

    # BJP
    if "BHARATIYA" in compact and (
        "JANATA" in compact or
        "JANATHA" in compact
    ):
        return "BJP"

    # BSP
    if "BAHUJANSAMAJ" in compact:
        return "BSP"

    # NTK
    if "NAAMTAMILAR" in compact:
        return "NTK"

    # PMK
    if "PATTALIMAKKAL" in compact:
        return "PMK"

    # AMMK
    if "AMMAMAKKALMUNNETRAKAZHAGAM" in compact:
        return "AMMK"

    # MNM
    if "MAKKALNEEDHIMAIAM" in compact:
        return "MNM"

    # CPI(M-L) Liberation
    if (
        "COMMUNISTPARTYOFINDIA" in compact
        and "MARXISTLENINIST" in compact
    ):
        return "CPI(ML)L"

    # Independent
    if "INDEPENDENT" in compact:
        return "IND"

    # Other parties remain cleaned
    return original
```

```python
def safe_ratio(numerator, denominator):

    numerator = pd.to_numeric(
        numerator,
        errors="coerce"
    )

    denominator = pd.to_numeric(
        denominator,
        errors="coerce"
    )

    return np.where(
        denominator > 0,
        numerator / denominator,
        np.nan
    )
```

```python
# ================================================================
# ELECTION ANALYTICS / PREDICTION 2026
# SRIPERUMBUDUR
# COMPLETE ML + DL PIPELINE
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
import sqlite3
import joblib
import warnings

from pathlib import Path

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

print("Starting complete project pipeline...")


# ================================================================
# 1. PROJECT FOLDERS
# ================================================================

CURRENT_DIR = Path.cwd()

DATA_DIR = CURRENT_DIR / "data"
RESULTS_DIR = CURRENT_DIR / "results"
MODELS_DIR = CURRENT_DIR / "models"
DATABASE_DIR = CURRENT_DIR / "database"
REPORTS_DIR = CURRENT_DIR / "reports"

for folder in [
    DATA_DIR,
    RESULTS_DIR,
    MODELS_DIR,
    DATABASE_DIR,
    REPORTS_DIR
]:
    folder.mkdir(exist_ok=True)

print("Folders ready")


# ================================================================
# 2. LOAD DATA AGAIN
# ================================================================

df2016 = pd.read_csv(
    DATA_DIR / "form20_2016_long.csv"
)

df2016_candidates = pd.read_csv(
    DATA_DIR / "form20_2016_candidates.csv"
)

df2021 = pd.read_csv(
    DATA_DIR / "form20_2021_long.csv"
)

df2021_candidates = pd.read_csv(
    DATA_DIR / "form20_2021_candidates.csv"
)

df2026 = pd.read_csv(
    DATA_DIR / "booth_2026_mla.csv"
)

print("\nData loaded")

print("2016:", df2016.shape)
print("2021:", df2021.shape)
print("2026:", df2026.shape)


# ================================================================
# 3. CLEAN TEXT
# ================================================================

def clean_text(value):

    if pd.isna(value):
        return ""

    value = str(value).upper().strip()

    value = value.replace(",", " ")

    value = re.sub(
        r"[^A-Z0-9&() ]+",
        " ",
        value
    )

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()


# ================================================================
# 4. PARTY STANDARDIZATION
# ================================================================

def canonical_party(value):

    original = clean_text(value)

    if original in ["", "0", "NAN"]:
        return "UNKNOWN"

    compact = re.sub(
        r"[^A-Z0-9]+",
        "",
        original
    )

    if "INDIANNATIONALCONGRESS" in compact:
        return "INC"

    if "ALLINDIAANNADRAVIDAMUNNETRAKAZHAGAM" in compact:
        return "AIADMK"

    if "ANNADRAVIDAMUNNETRAKAZHAGAM" in compact:
        return "AIADMK"

    if "DRAVIDAMUNNETRAKAZHAGAM" in compact:
        return "DMK"

    if "BAHUJANSAMAJ" in compact:
        return "BSP"

    if (
        "BHARATIYA" in compact
        and
        ("JANATA" in compact or "JANATHA" in compact)
    ):
        return "BJP"

    if "NAAMTAMILAR" in compact:
        return "NTK"

    if "PATTALIMAKKAL" in compact:
        return "PMK"

    if "AMMAMAKKALMUNNETRAKAZHAGAM" in compact:
        return "AMMK"

    if "MAKKALNEEDHIMAIAM" in compact:
        return "MNM"

    if (
        "COMMUNISTPARTYOFINDIA" in compact
        and
        "MARXISTLENINIST" in compact
    ):
        return "CPI(ML)L"

    if "INDEPENDENT" in compact:
        return "IND"

    return original


# ================================================================
# 5. CLEAN HISTORICAL TEXT
# ================================================================

for df in [df2016, df2021]:

    df["candidate_clean"] = (
        df["candidate"]
        .apply(clean_text)
    )

    df["party_clean"] = (
        df["party"]
        .apply(canonical_party)
    )


# ================================================================
# 6. NUMERIC CONVERSION
# ================================================================

for col in [
    "ps_no",
    "votes",
    "valid_votes",
    "total_votes"
]:

    df2016[col] = pd.to_numeric(
        df2016[col],
        errors="coerce"
    )

    df2021[col] = pd.to_numeric(
        df2021[col],
        errors="coerce"
    )


# ================================================================
# 7. CREATE SUFFIX KEY
# ================================================================

df2016["suffix_clean"] = (
    df2016["ps_suffix"]
    .fillna("MAIN")
    .astype(str)
    .str.strip()
    .str.upper()
)

df2021["suffix_clean"] = (
    df2021["ps_suffix"]
    .fillna("MAIN")
    .astype(str)
    .str.strip()
    .str.upper()
)


# ================================================================
# 8. CORRECT 2016 BOOTH TOTALS
# ================================================================

party_votes_2016 = (
    df2016
    .groupby(
        ["ps_no", "party_clean"],
        as_index=False
    )
    .agg(
        votes_2016=("votes", "sum"),
        candidate_count_2016=("candidate_clean", "nunique")
    )
)


booth_totals_2016_suffix = (
    df2016
    .groupby(
        ["ps_no", "suffix_clean"],
        as_index=False
    )
    .agg(
        valid_votes=("valid_votes", "max"),
        total_votes=("total_votes", "max")
    )
)


booth_totals_2016 = (
    booth_totals_2016_suffix
    .groupby(
        "ps_no",
        as_index=False
    )
    .agg(
        valid_votes_2016=("valid_votes", "sum"),
        total_votes_2016=("total_votes", "sum")
    )
)


hist2016 = pd.merge(
    party_votes_2016,
    booth_totals_2016,
    on="ps_no",
    how="left"
)


hist2016["vote_share_2016"] = np.where(

    hist2016["valid_votes_2016"] > 0,

    hist2016["votes_2016"]
    /
    hist2016["valid_votes_2016"],

    np.nan
)


hist2016["valid_ratio_2016"] = np.where(

    hist2016["total_votes_2016"] > 0,

    hist2016["valid_votes_2016"]
    /
    hist2016["total_votes_2016"],

    np.nan
)


# ================================================================
# 9. CORRECT 2021 BOOTH TOTALS
# ================================================================

party_votes_2021 = (
    df2021
    .groupby(
        ["ps_no", "party_clean"],
        as_index=False
    )
    .agg(
        votes_2021=("votes", "sum"),
        candidate_count_2021=("candidate_clean", "nunique")
    )
)


booth_totals_2021_suffix = (
    df2021
    .groupby(
        ["ps_no", "suffix_clean"],
        as_index=False
    )
    .agg(
        valid_votes=("valid_votes", "max"),
        total_votes=("total_votes", "max")
    )
)


booth_totals_2021 = (
    booth_totals_2021_suffix
    .groupby(
        "ps_no",
        as_index=False
    )
    .agg(
        valid_votes_2021=("valid_votes", "sum"),
        total_votes_2021=("total_votes", "sum")
    )
)


hist2021 = pd.merge(
    party_votes_2021,
    booth_totals_2021,
    on="ps_no",
    how="left"
)


hist2021["vote_share_2021"] = np.where(

    hist2021["valid_votes_2021"] > 0,

    hist2021["votes_2021"]
    /
    hist2021["valid_votes_2021"],

    np.nan
)


hist2021["valid_ratio_2021"] = np.where(

    hist2021["total_votes_2021"] > 0,

    hist2021["valid_votes_2021"]
    /
    hist2021["total_votes_2021"],

    np.nan
)


# ================================================================
# 10. VALIDATE VOTE SHARES
# ================================================================

print("\n2016 vote share max:",
      hist2016["vote_share_2016"].max())

print("2021 vote share max:",
      hist2021["vote_share_2021"].max())

print(
    "2016 shares > 1:",
    (hist2016["vote_share_2016"] > 1).sum()
)

print(
    "2021 shares > 1:",
    (hist2021["vote_share_2021"] > 1).sum()
)


# Remove invalid rows if any remain

hist2016 = hist2016[
    hist2016["vote_share_2016"].between(0, 1)
].copy()

hist2021 = hist2021[
    hist2021["vote_share_2021"].between(0, 1)
].copy()


# ================================================================
# 11. MERGE 2016 AND 2021
# ================================================================

model_df = pd.merge(
    hist2016,
    hist2021,
    on=[
        "ps_no",
        "party_clean"
    ],
    how="inner"
)


model_df = model_df.replace(
    [np.inf, -np.inf],
    np.nan
)


model_df = model_df.dropna(
    subset=[
        "vote_share_2016",
        "vote_share_2021"
    ]
)


print(
    "\nMatched historical rows:",
    model_df.shape
)


# ================================================================
# 12. FEATURE ENGINEERING
# ================================================================

model_df["vote_share_change"] = (
    model_df["vote_share_2021"]
    -
    model_df["vote_share_2016"]
)


model_df["vote_count_change"] = (
    model_df["votes_2021"]
    -
    model_df["votes_2016"]
)


model_df.to_csv(
    RESULTS_DIR / "historical_training_dataset.csv",
    index=False
)


# ================================================================
# 13. INPUT FEATURES / TARGET
# ================================================================

numeric_features = [

    "ps_no",

    "votes_2016",

    "valid_votes_2016",

    "total_votes_2016",

    "vote_share_2016",

    "valid_ratio_2016",

    "candidate_count_2016"

]


categorical_features = [

    "party_clean"

]


target = "vote_share_2021"


X = model_df[
    numeric_features
    +
    categorical_features
].copy()


y = model_df[target].copy()


print("\nFeatures:", numeric_features + categorical_features)
print("Target:", target)


# ================================================================
# 14. TRAIN / TEST SPLIT BY BOOTH
# ================================================================

groups = model_df["ps_no"]


splitter = GroupShuffleSplit(

    n_splits=1,

    test_size=0.20,

    random_state=42
)


train_index, test_index = next(

    splitter.split(
        X,
        y,
        groups
    )
)


X_train = X.iloc[train_index].copy()

X_test = X.iloc[test_index].copy()

y_train = y.iloc[train_index].copy()

y_test = y.iloc[test_index].copy()


print("\nTraining rows:", len(X_train))
print("Testing rows:", len(X_test))


# ================================================================
# 15. PREPROCESSOR
# ================================================================

def create_preprocessor(scale_numeric=False):

    numeric_steps = [

        (
            "imputer",

            SimpleImputer(
                strategy="median"
            )
        )

    ]


    if scale_numeric:

        numeric_steps.append(

            (
                "scaler",

0                StandardScaler()
            )

        )


    numeric_pipeline = Pipeline(
        numeric_steps
    )


    categorical_pipeline = Pipeline(

        steps=[

            (
                "imputer",

                SimpleImputer(
                    strategy="most_frequent"
                )
            ),

            (
                "onehot",

                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )

        ]

    )


    preprocessor = ColumnTransformer(

        transformers=[

            (
                "numeric",

                numeric_pipeline,

                numeric_features
            ),

            (
                "categorical",

                categorical_pipeline,

                categorical_features
            )

        ]

    )


    return preprocessor


# ================================================================
# 16. EVALUATION FUNCTION
# ================================================================

def evaluate_model(
    model_name,
    actual,
    prediction
):

    prediction = np.asarray(
        prediction
    ).reshape(-1)


    prediction = np.clip(
        prediction,
        0,
        1
    )


    mae = mean_absolute_error(
        actual,
        prediction
    )


    rmse = np.sqrt(
        mean_squared_error(
            actual,
            prediction
        )
    )


    r2 = r2_score(
        actual,
        prediction
    )


    print("\n" + "=" * 50)

    print(model_name)

    print("=" * 50)

    print("MAE :", round(mae, 5))

    print("RMSE:", round(rmse, 5))

    print("R²  :", round(r2, 5))


    return {

        "Model": model_name,

        "MAE": mae,

        "RMSE": rmse,

        "R2": r2

    }


# ================================================================
# 17. LINEAR REGRESSION
# ================================================================

linear_model = Pipeline([

    (
        "preprocessor",

        create_preprocessor(False)
    ),

    (
        "model",

        LinearRegression()
    )

])


linear_model.fit(
    X_train,
    y_train
)


linear_prediction = linear_model.predict(
    X_test
)


linear_result = evaluate_model(

    "Linear Regression",

    y_test,

    linear_prediction
)


joblib.dump(

    linear_model,

    MODELS_DIR / "linear_regression.joblib"
)


# ================================================================
# 18. RANDOM FOREST
# ================================================================

rf_model = Pipeline([

    (
        "preprocessor",

        create_preprocessor(False)
    ),

    (
        "model",

        RandomForestRegressor(

            n_estimators=300,

            max_depth=10,

            min_samples_leaf=2,

            random_state=42,

            n_jobs=-1
        )
    )

])


rf_model.fit(
    X_train,
    y_train
)


rf_prediction = rf_model.predict(
    X_test
)


rf_result = evaluate_model(

    "Random Forest",

    y_test,

    rf_prediction
)


joblib.dump(

    rf_model,

    MODELS_DIR / "random_forest.joblib"
)


# ================================================================
# 19. GRADIENT BOOSTING
# ================================================================

gb_model = Pipeline([

    (
        "preprocessor",

        create_preprocessor(False)
    ),

    (
        "model",

        GradientBoostingRegressor(

            n_estimators=250,

            learning_rate=0.04,

            max_depth=3,

            random_state=42
        )
    )

])


gb_model.fit(
    X_train,
    y_train
)


gb_prediction = gb_model.predict(
    X_test
)


gb_result = evaluate_model(

    "Gradient Boosting",

    y_test,

    gb_prediction
)


joblib.dump(

    gb_model,

    MODELS_DIR / "gradient_boosting.joblib"
)


# ================================================================
# 20. XGBOOST
# ================================================================

xgb_result = None
xgb_prediction = None


try:

    from xgboost import XGBRegressor


    xgb_model = Pipeline([

        (
            "preprocessor",

            create_preprocessor(False)
        ),

        (
            "model",

            XGBRegressor(

                n_estimators=300,

                learning_rate=0.03,

                max_depth=4,

                subsample=0.85,

                colsample_bytree=0.85,

                objective="reg:squarederror",

                random_state=42,

                n_jobs=-1
            )
        )

    ])


    xgb_model.fit(
        X_train,
        y_train
    )


    xgb_prediction = xgb_model.predict(
        X_test
    )


    xgb_result = evaluate_model(

        "XGBoost",

        y_test,

        xgb_prediction
    )


    joblib.dump(

        xgb_model,

        MODELS_DIR / "xgboost.joblib"
    )


except Exception as e:

    print("\nXGBoost skipped:", e)


# ================================================================
# 21. DEEP LEARNING PREPROCESSING
# ================================================================

dl_result = None
dl_prediction = None


try:

    import tensorflow as tf

    from tensorflow.keras.models import Sequential

    from tensorflow.keras.layers import (
        Input,
        Dense,
        Dropout
    )

    from tensorflow.keras.callbacks import EarlyStopping


    tf.random.set_seed(42)

    np.random.seed(42)


    dl_preprocessor = create_preprocessor(
        True
    )


    X_train_dl = dl_preprocessor.fit_transform(
        X_train
    )


    X_test_dl = dl_preprocessor.transform(
        X_test
    )


    joblib.dump(

        dl_preprocessor,

        MODELS_DIR /
        "deep_learning_preprocessor.joblib"
    )


    print(
        "\nDeep Learning input:",
        X_train_dl.shape
    )


# ================================================================
# 22. MLP MODEL
# ================================================================

    dl_model = Sequential([

        Input(
            shape=(X_train_dl.shape[1],)
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            0.15
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="sigmoid"
        )

    ])


    dl_model.compile(

        optimizer=
        tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),

        loss="mse",

        metrics=["mae"]
    )


    early_stop = EarlyStopping(

        monitor="val_loss",

        patience=15,

        restore_best_weights=True
    )


    history = dl_model.fit(

        X_train_dl,

        y_train.to_numpy(),

        validation_split=0.20,

        epochs=150,

        batch_size=32,

        callbacks=[
            early_stop
        ],

        verbose=1
    )


    dl_prediction = (

        dl_model.predict(
            X_test_dl,
            verbose=0
        )

        .flatten()
    )


    dl_result = evaluate_model(

        "MLP Deep Learning",

        y_test,

        dl_prediction
    )


    dl_model.save(

        MODELS_DIR /
        "mlp_model.keras"
    )


    pd.DataFrame(
        history.history
    ).to_csv(

        RESULTS_DIR /
        "deep_learning_history.csv",

        index=False
    )


except Exception as e:

    print(
        "\nDeep Learning skipped:",
        e
    )


# ================================================================
# 23. MODEL COMPARISON
# ================================================================

results = [

    linear_result,

    rf_result,

    gb_result

]


if xgb_result is not None:

    results.append(
        xgb_result
    )


if dl_result is not None:

    results.append(
        dl_result
    )


results_df = pd.DataFrame(
    results
)


results_df = (

    results_df

    .sort_values(
        "RMSE"
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nFINAL HISTORICAL MODEL RESULTS"
)

display(
    results_df
)


results_df.to_csv(

    RESULTS_DIR /
    "model_comparison.csv",

    index=False
)


# ================================================================
# 24. MODEL COMPARISON GRAPH
# ================================================================

plt.figure(
    figsize=(10, 5)
)


plt.bar(

    results_df["Model"],

    results_df["RMSE"]
)


plt.ylabel(
    "RMSE"
)

plt.title(
    "Historical Model RMSE Comparison"
)

plt.xticks(
    rotation=25
)

plt.tight_layout()


plt.savefig(

    RESULTS_DIR /
    "model_comparison.png",

    dpi=150
)


plt.show()


# ================================================================
# 25. SAVE HISTORICAL PREDICTIONS
# ================================================================

prediction_output = pd.DataFrame({

    "actual_2021":
        y_test.reset_index(
            drop=True
        ),

    "linear_regression":
        np.clip(
            linear_prediction,
            0,
            1
        ),

    "random_forest":
        np.clip(
            rf_prediction,
            0,
            1
        ),

    "gradient_boosting":
        np.clip(
            gb_prediction,
            0,
            1
        )

})


if xgb_prediction is not None:

    prediction_output["xgboost"] = np.clip(
        xgb_prediction,
        0,
        1
    )


if dl_prediction is not None:

    prediction_output["deep_learning"] = np.clip(
        dl_prediction,
        0,
        1
    )


prediction_output.to_csv(

    RESULTS_DIR /
    "historical_predictions.csv",

    index=False
)


# ================================================================
# 26. 2026 DATA CLEANING
# ================================================================

booth2026 = df2026.copy()


if "row_type" in booth2026.columns:

    booth2026 = booth2026[

        booth2026[
            "row_type"
        ]

        .astype(str)

        .str.lower()

        .eq("booth")

    ].copy()


for col in [

    "booth_no",

    "polled",

    "inc",

    "admk",

    "tvk"

]:

    booth2026[col] = pd.to_numeric(

        booth2026[col],

        errors="coerce"
    )


booth2026 = booth2026.dropna(

    subset=[
        "booth_no",
        "polled"
    ]
)


booth2026 = booth2026[

    booth2026["polled"] > 0

].copy()


booth2026["booth_no"] = (

    booth2026["booth_no"]

    .astype(int)
)


# ================================================================
# 27. 2026 BOOTH SHARE ANALYSIS
# ================================================================

booth2026["inc_share"] = (

    booth2026["inc"]

    /

    booth2026["polled"]
)


booth2026["admk_share"] = (

    booth2026["admk"]

    /

    booth2026["polled"]
)


booth2026["tvk_share"] = (

    booth2026["tvk"]

    /

    booth2026["polled"]
)


booth2026["listed_votes"] = (

    booth2026[[
        "inc",
        "admk",
        "tvk"
    ]]

    .sum(axis=1)
)


booth2026["other_votes"] = (

    booth2026["polled"]

    -

    booth2026["listed_votes"]

).clip(lower=0)


booth2026["other_share"] = (

    booth2026["other_votes"]

    /

    booth2026["polled"]
)


booth2026.to_csv(

    RESULTS_DIR /
    "booth_2026_analysis.csv",

    index=False
)


print(
    "\n2026 booth data:"
)

display(

    booth2026[
        [
            "booth_no",
            "booth_name",
            "polled",
            "inc_share",
            "admk_share",
            "tvk_share"
        ]
    ]
    .head(20)
)


# ================================================================
# 28. 2026 PARTY TOTAL SUMMARY
# ================================================================

total_polled = booth2026[
    "polled"
].sum()


summary2026 = pd.DataFrame({

    "party": [
        "INC",
        "AIADMK",
        "TVK"
    ],

    "votes": [

        booth2026["inc"].sum(),

        booth2026["admk"].sum(),

        booth2026["tvk"].sum()

    ]

})


summary2026["share"] = (

    summary2026["votes"]

    /

    total_polled
)


print(
    "\n2026 supplied data summary"
)

display(
    summary2026
)


summary2026.to_csv(

    RESULTS_DIR /
    "2026_party_summary.csv",

    index=False
)


# ================================================================
# 29. 2026 BOOTH GRAPH
# ================================================================

graph_df = (

    booth2026

    .sort_values(
        "booth_no"
    )

    .head(50)
)


plt.figure(
    figsize=(14, 6)
)


plt.plot(
    graph_df["booth_no"],
    graph_df["inc_share"],
    label="INC"
)


plt.plot(
    graph_df["booth_no"],
    graph_df["admk_share"],
    label="AIADMK"
)


plt.plot(
    graph_df["booth_no"],
    graph_df["tvk_share"],
    label="TVK"
)


plt.title(
    "2026 Supplied Booth-Level Vote Share"
)

plt.xlabel(
    "Booth Number"
)

plt.ylabel(
    "Share"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    RESULTS_DIR /
    "2026_booth_share.png",

    dpi=150
)


plt.show()


# ================================================================
# 30. SQLITE DATABASE
# ================================================================

db_path = (

    DATABASE_DIR /
    "election2026.db"
)


conn = sqlite3.connect(
    db_path
)


df2016.to_sql(

    "raw_2016",

    conn,

    if_exists="replace",

    index=False
)


df2021.to_sql(

    "raw_2021",

    conn,

    if_exists="replace",

    index=False
)


df2026.to_sql(

    "raw_2026",

    conn,

    if_exists="replace",

    index=False
)


hist2016.to_sql(

    "clean_2016",

    conn,

    if_exists="replace",

    index=False
)


hist2021.to_sql(

    "clean_2021",

    conn,

    if_exists="replace",

    index=False
)


model_df.to_sql(

    "historical_training",

    conn,

    if_exists="replace",

    index=False
)


results_df.to_sql(

    "model_results",

    conn,

    if_exists="replace",

    index=False
)


booth2026.to_sql(

    "booth_2026_analysis",

    conn,

    if_exists="replace",

    index=False
)


conn.close()


print(
    "\nDatabase created:",
    db_path
)


# ================================================================
# 31. FINAL REPORT TXT
# ================================================================

best_model = results_df.iloc[0]


report = f"""
ELECTION ANALYTICS 2026
SRIPERUMBUDUR

========================================

Historical data:
2016 -> 2021

Historical matched observations:
{len(model_df)}

Training observations:
{len(X_train)}

Testing observations:
{len(X_test)}

----------------------------------------

MODEL COMPARISON

{results_df.to_string(index=False)}

----------------------------------------

LOWEST HISTORICAL RMSE MODEL:

{best_model['Model']}

MAE:
{best_model['MAE']:.5f}

RMSE:
{best_model['RMSE']:.5f}

R2:
{best_model['R2']:.5f}

----------------------------------------

DEEP LEARNING CONFIGURATION

Model:
Multilayer Perceptron

Hidden Layer 1:
64 neurons
Activation = ReLU

Dropout = 20%

Hidden Layer 2:
32 neurons
Activation = ReLU

Dropout = 15%

Hidden Layer 3:
16 neurons
Activation = ReLU

Output:
1 neuron
Sigmoid activation

Optimizer:
Adam

Learning Rate:
0.001

Loss:
Mean Squared Error

Batch Size:
32

Maximum Epochs:
150

Early Stopping:
Enabled

----------------------------------------

LIMITATIONS

The model is evaluated as a historical
2016-to-2021 backtest.

Polling-station numbering and boundaries
may change between elections.

Assembly and local-body elections can
involve different wards, candidates,
alliances and turnout.

2026 supplied booth data is analysed
descriptively and should not be treated
as certainty about a future election result.

========================================
"""


report_path = (

    REPORTS_DIR /
    "final_report.txt"
)


with open(

    report_path,

    "w",

    encoding="utf-8"

) as f:

    f.write(report)


print(report)


# ================================================================
# 32. FINAL OUTPUT
# ================================================================

print("\n" + "=" * 70)

print(
    "PROJECT PIPELINE COMPLETED"
)

print("=" * 70)


print(
    "\nResults folder:"
)

for file in RESULTS_DIR.iterdir():

    print(
        "-",
        file.name
    )


print(
    "\nModels folder:"
)

for file in MODELS_DIR.iterdir():

    print(
        "-",
        file.name
    )


print(
    "\nDatabase:"
)

print(
    "-",
    db_path
)


print(
    "\nReport:"
)

print(
    "-",
    report_path
)
```
